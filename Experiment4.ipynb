```python
# 1. Install dependency
!pip install -q python-docx

# 2. Extract Experiment 4 content securely
import os
import docx
import cv2
import matplotlib.pyplot as plt
import urllib.request
import numpy as np

# Search for the document in the current directory and subdirectories
target_file = 'experiment lab 4 5 6.docx'
file_path = None

for root, dirs, files in os.walk('/content'):
    for file in files:
        if file.lower() == target_file.lower():
            file_path = os.path.join(root, file)
            break

# If the document is found, read it; otherwise, use the known code from the document
if file_path and os.path.exists(file_path):
    print(f"Found document at: {file_path}")
    doc = docx.Document(file_path)
    full_text = [para.text for para in doc.paragraphs]
    docx_text = '\n'.join(full_text)
    exp5_idx = docx_text.lower().find('experiment 5')
    exp4_code = docx_text[:exp5_idx] if exp5_idx != -1 else docx_text
else:
    print("Document file not found physically. Using the previously extracted Experiment 4 code:")
    exp4_code = """from google.colab import files
import cv2
import matplotlib.pyplot as plt
# ... (Histogram Equalization Experiment) ..."""

print("=== EXPERIMENT 4 CODE TO RUN ===")
print(exp4_code.strip())
print("================================\n")

# 3. Download a sample image to execute the code without requiring manual uploads
image_url = "https://raw.githubusercontent.com/opencv/opencv/master/samples/data/lena.jpg"
sample_path = "sample_image.jpg"
try:
    urllib.request.urlretrieve(image_url, sample_path)
    print("Downloaded sample image successfully.")
except Exception as e:
    # Create a synthetic image as a fallback
    print("Generating a synthetic grayscale image.")
    synthetic = np.zeros((400, 400), dtype=np.uint8)
    for i in range(400):
        synthetic[:, i] = int((i / 400.0) * 255)
    cv2.imwrite(sample_path, synthetic)

# 4. Run Experiment 4 Logic (Histogram Equalization)
image = cv2.imread(sample_path)
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
equalized = cv2.equalizeHist(gray)

# Show Original vs Enhanced
plt.figure(figsize=(12, 5))
plt.subplot(1, 2, 1)
plt.imshow(gray, cmap="gray")
plt.title("Original Grayscale Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(equalized, cmap="gray")
plt.title("Histogram Equalized Image")
plt.axis("off")
plt.show()

# Show Histograms
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
plt.hist(gray.ravel(), 256, [0, 256], color='gray')
plt.title("Original Histogram")
plt.xlabel("Pixel Value")
plt.ylabel("Frequency")

plt.subplot(1, 2, 2)
plt.hist(equalized.ravel(), 256, [0, 256], color='blue')
plt.title("Equalized Histogram")
plt.xlabel("Pixel Value")
plt.ylabel("Frequency")
plt.show()
```

```python
import docx
import re

def extract_document_sections(file_path):
    doc = docx.Document(file_path)
    full_text = []
    for para in doc.paragraphs:
        full_text.append(para.text)

    # Also read tables
    for table in doc.tables:
        for row in table.rows:
            row_text = [cell.text.strip() for cell in row.cells if cell.text.strip()]
            if row_text:
                full_text.append(" | ".join(row_text))

    return '\n'.join(full_text)

text = extract_document_sections('/content/experiment lab 4 5 6.docx')

# Let's locate Experiment 4 specifically
print("=== Document Length:", len(text), "characters ===")

# Try to find headers related to Experiment 4, 5, 6
pattern = re.compile(r'(experiment\s*(?:no\.?\s*)?[456]|lab\s*[456])', re.IGNORECASE)
matches = list(pattern.finditer(text))

if matches:
    for i, match in enumerate(matches):
        start = match.start()
        end = matches[i+1].start() if i + 1 < len(matches) else len(text)
        print(f"\n--- Section found: {match.group(0)} ---")
        print(text[start:start+1200] + "\n...[Truncated]...")
else:
    print("Could not find specific 'Experiment 4' headers, showing the first 3000 characters:")
    print(text[:3000])
```
